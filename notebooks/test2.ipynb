```python
import numpy as np
import pandas as pd
```

```python
bbb = pd.read_csv('data/ipl_dataset_2008-2026/ball_by_ball_data.csv')
bbb_2026 = pd.read_csv('data/ipl_ballbyball_2026_sample.csv')
```

```python
bbb['ball_in_innings'] = bbb.apply(lambda x: 0 if (x['is_wide_ball'] or x['is_no_ball']) else 1, axis=1)
bbb['ball_in_innings'] = bbb.groupby(['match_id', 'innings']).agg(
    {
        "ball_in_innings":"cumsum"
    }
)
bbb['ball_in_innings'] = bbb['ball_in_innings'].astype(int)
bbb
```
